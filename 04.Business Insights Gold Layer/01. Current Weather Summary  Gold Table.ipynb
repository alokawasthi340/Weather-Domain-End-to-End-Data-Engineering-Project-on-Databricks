```python
%sql
create schema if not exists weather.gold;
```

```python
source_table= "weather.silver.current_weather"
target_table= "weather.gold.current_weather_summary"
```

```python
from pyspark.sql.functions import *

current_weather_summary_df=(
    spark.read.table(source_table)
    .select(
        col("city"),
        col("temperature"),
        col("feels_like"),
        col("humidity"),
        col("pressure"),
        col("visibility"),
        col("wind_speed"),
        col("wind_direction"),
        col("country"),
        col("latitude"),
        col("longitude"),
    )
    .withColumn("load_dt_tm", current_timestamp())
    .withColumn("update_dt_tm", current_timestamp())
)
display(current_weather_summary_df)
```

```python
(
    current_weather_summary_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(target_table)
)
```
