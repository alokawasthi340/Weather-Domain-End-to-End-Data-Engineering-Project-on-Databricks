```python
source_table= "weather.silver.current_weather"
target_table= "weather.gold.air_quality_summary"
```

```python
from pyspark.sql.functions import *

air_quality_summary_df = spark.read.table(source_table).select(
    col("city"), col("co"), col("no2"), col("o3"), col("pm10"), col("pm25")
)
display(air_quality_summary_df)
```

```python
air_quality_summary_df.write.format("delta").mode("overwrite").saveAsTable(target_table)
```
