```python
%sql
create schema if not exists weather.gold;
```

```python
source_table= "weather.silver.forecast_weather"
target_table= "weather.gold.forecast_weather_summary"
```

```python
from pyspark.sql.functions import *

forecast_weather_df=(
    spark.read.table(source_table)
    .select(
        col("city"),
        col("forecast_date"),
        col("avgtemp"),
        col("maxtemp"),
        col("mintemp"),
        col("sunhour"),
        col("uv_index"),
        col("co"),
        col("pm10"),
    )
    .withColumn("load_dt_tm", current_timestamp())
    .withColumn("update_dt_tm",current_timestamp())
)
display(forecast_weather_df)
```

```python
(forecast_weather_df.write.format("delta").mode("overwrite").saveAsTable(target_table))
```
