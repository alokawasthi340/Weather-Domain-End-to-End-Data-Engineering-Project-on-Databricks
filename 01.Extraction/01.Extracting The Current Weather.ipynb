```python
locations=["Mumbai" ,"Bangalore" ,"Hyderabad" ,"Chennai" ,"Kolkata" ,"Pune" ,"Ahmedabad"]
```

```python
import requests
import urllib3
import json
import time
import os
from datetime import date
import logging

urllib3.disable_warnings(urllib3.exceptions.InsecureRequestWarning)

logging.basicConfig(level=logging.INFO)
logger = logging.getLogger("weather_logger")

def current_weather(locations: list):
    # Ensure location names are spelled correctly and supported by the API
    api_key = dbutils.secrets.get("weather", "weather-api-key")
    url = "https://api.weatherstack.com/current"
    headers = {"Accept": "application/json"}
    responses = []
    for location in locations:
        try:
            querystring = {"query": location, "access_key": api_key}
            response = requests.get(url, headers=headers, params=querystring, verify=False, timeout=10)
            result = response.json()
            responses.append({"city": location, "data": result})
        except requests.RequestException as exc:
            error_message = str(exc)
            responses.append({"city": location, "error": error_message})
        except ValueError as exc:
            responses.append({"city": location, "error": f"Invalid JSON: {exc}"})
        time.sleep(2)
    try:
        volume_path = "/Volumes/weather/landing/files/current_weather"
        file_path = f"{volume_path}/current_weather_{date.today()}.json"
        with open(file_path, "w", encoding="utf-8") as f:
            json.dump(responses, f, ensure_ascii=False, indent=2)
    except OSError as exc:
        logger.error(f"Failed to write {file_path}.json: {exc}")
```
