```python
%sql
create schema if not exists weather.bronze;
```

```python
# Define your volume path and table names
volume_path = "/Volumes/weather/landing/files/forecast_weather"
target_table= "weather.bronze.forecast_weather"
```

```python
from pyspark.sql.functions import col, current_timestamp

frocast_df = (
    spark.read
    .option("multiline", "true")
    .json(volume_path)
    .withColumn("source_file", col("_metadata.file_name"))
    .withColumn("source_filepath", col("_metadata.file_path"))
    .withColumn("ingestion_timestamp", current_timestamp())
)

display(frocast_df)
```

```python
frocast_df.write.format("delta").mode("append").saveAsTable(target_table)
```
